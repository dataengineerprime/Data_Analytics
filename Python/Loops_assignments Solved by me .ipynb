```python
##Assignment 4
num = float(input("Enter any number: "))

if num == 0:
    print(f"{num} is zero")
elif num <0:
    print(f"{num} is a negative number") 
elif num > 0 and num.is_integer():
    if num % 2 == 0:
        print(f"{num} is an even number")
    else:
        print(f"{num} is an odd number")   
else:
    print(f"{num} is a positive number")
    
```

```python
#Assignment 13
number = int(input("Enter any number: "))
sum_of_digit = 0
while number > 0:
    digit = number % 10
    sum_of_digit += digit
    number = number // 10
print(f"The sum of digits is {sum_of_digit}")
```

```python
#Assignment 14
number = int(input("Enter a number: "))

is_prime = True

if number <= 1:
    is_prime = False
else:
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")

    
```

```python
##Assignment 15
n = int(input("Enter the number of Fibonacci numbers to print: "))

a, b = 0, 1
count = 0

while count < n:
    print(a)
    a, b = b, a + b
    count += 1
```
